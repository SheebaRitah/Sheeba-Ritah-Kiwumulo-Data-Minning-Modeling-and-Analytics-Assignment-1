```python
#import libraries
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

warnings.filterwarnings("ignore")

# Load the dataset

# The dataset is stored in the Coursework folder
DATA_PATH = r"D:\Sheeba\Coursework\Concrete_Data.csv"

# Check whether the dataset exists
if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {DATA_PATH}\n"
        "Confirm that Concrete_Data.csv is in D:\\Sheeba\\Coursework."
    )

# Read the CSV file
df = pd.read_csv(DATA_PATH)

# Remove unwanted spaces from column names
df.columns = df.columns.str.strip()

print("DATASET LOADED SUCCESSFULLY")
print("Dataset shape:", df.shape)

display(df.head())

#explore the data

print("\nDATASET INFORMATION")
df.info()

print("\nSUMMARY STATISTICS")
display(df.describe().round(2))

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nDUPLICATE RECORDS")
print(df.duplicated().sum())

#Data preparation

# Remove exact duplicate records
df = df.drop_duplicates().reset_index(drop=True)

print("\nDataset shape after removing duplicates:", df.shape)

# Identify the target column
target_matches = [
    col for col in df.columns
    if "compressive strength" in col.lower()
]

if len(target_matches) != 1:
    raise ValueError(
        "Could not identify the compressive strength target column. "
        f"Available columns: {list(df.columns)}"
    )

TARGET = target_matches[0]

# Separate input variables and target variable
X = df.drop(columns=[TARGET])
y = df[TARGET]

# Ensure all variables are numeric
X = X.apply(pd.to_numeric, errors="coerce")
y = pd.to_numeric(y, errors="coerce")

# Remove rows with missing target values
valid_rows = y.notna()
X = X.loc[valid_rows].reset_index(drop=True)
y = y.loc[valid_rows].reset_index(drop=True)

print("\nINPUT VARIABLES")
print(list(X.columns))

print("\nTARGET VARIABLE")
print(TARGET)

print("\nFinal dataset shape:", X.shape)

#data exploration

# Plot distributions of all variables
df.hist(figsize=(15, 12), bins=20)
plt.suptitle("Distributions of Concrete Dataset Variables", fontsize=15)
plt.tight_layout()
plt.show()

# Correlation heatmap using matplotlib
correlation = df.corr(numeric_only=True)

plt.figure(figsize=(11, 8))
plt.imshow(correlation, cmap="coolwarm", aspect="auto")
plt.colorbar(label="Correlation")
plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=90
)
plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)
plt.title("Correlation Matrix of Concrete Dataset Variables")
plt.tight_layout()
plt.show()

# Relationship between cement and compressive strength
cement_column = next(
    (col for col in X.columns if "cement" in col.lower()),
    None
)

if cement_column is not None:
    plt.figure(figsize=(8, 5))
    plt.scatter(X[cement_column], y, alpha=0.5)
    plt.xlabel("Cement (kg per cubic metre)")
    plt.ylabel("Compressive Strength (MPa)")
    plt.title("Cement Content vs Concrete Compressive Strength")
    plt.tight_layout()
    plt.show()


# Training and testing data

# Use an 80% training and 20% testing split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTRAINING AND TESTING DATA")
print("Training records:", len(X_train))
print("Testing records:", len(X_test))

# Establishing the baseline

# The baseline predicts the mean training strength for every test record
baseline_prediction = np.full(
    len(y_test),
    y_train.mean()
)

baseline_mae = mean_absolute_error(
    y_test, baseline_prediction
)

baseline_rmse = np.sqrt(
    mean_squared_error(y_test, baseline_prediction)
)

baseline_r2 = r2_score(
    y_test, baseline_prediction
)

print("\nBASELINE PERFORMANCE")
print("MAE:", round(baseline_mae, 4))
print("RMSE:", round(baseline_rmse, 4))
print("R2:", round(baseline_r2, 4))

# Building the learning models

# Median imputation is included in each pipeline.
# Standardisation is applied to Linear Regression.
# Tree-based models do not require standardisation.

models = {
    "Linear Regression": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("model", LinearRegression())
    ]),

    "Random Forest": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", RandomForestRegressor(
            n_estimators=300,
            random_state=42,
            n_jobs=-1
        ))
    ]),

    "HistGradientBoosting": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", HistGradientBoostingRegressor(
            max_iter=200,
            learning_rate=0.08,
            random_state=42
        ))
    ])
}

# 9. Training and evaluating the models-

results = []
predictions = {}

for model_name, model in models.items():

    print(f"\nTraining {model_name}...")

    # Train the model
    model.fit(X_train, y_train)

    # Predict compressive strength on unseen test data
    y_pred = model.predict(X_test)

    predictions[model_name] = y_pred

    # Calculate performance metrics
    mae = mean_absolute_error(y_test, y_pred)

    rmse = np.sqrt(
        mean_squared_error(y_test, y_pred)
    )

    r2 = r2_score(y_test, y_pred)

    results.append({
        "Model": model_name,
        "MAE (MPa)": mae,
        "RMSE (MPa)": rmse,
        "R2 Score": r2
    })

# Include baseline in comparison
results.append({
    "Model": "Mean Baseline",
    "MAE (MPa)": baseline_mae,
    "RMSE (MPa)": baseline_rmse,
    "R2 Score": baseline_r2
})

results_df = pd.DataFrame(results)

# Rank models by RMSE (lower is better)
results_df = results_df.sort_values(
    by="RMSE (MPa)"
).reset_index(drop=True)

print("\nMODEL PERFORMANCE COMPARISON")
display(results_df.round(4))

# Identifying the best model

# The best model has the lowest test RMSE
best_model_name = min(
    models,
    key=lambda name: np.sqrt(
        mean_squared_error(
            y_test,
            predictions[name]
        )
    )
)

best_model = models[best_model_name]
best_predictions = predictions[best_model_name]

print("\nBEST PERFORMING MODEL")
print(best_model_name)

# 11. Visual model

# Compare RMSE values
plt.figure(figsize=(10, 5))

plt.bar(
    results_df["Model"],
    results_df["RMSE (MPa)"]
)

plt.ylabel("RMSE (MPa)")
plt.xlabel("Machine Learning Model")
plt.title("Comparison of Model RMSE")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

# Compare actual and predicted values
plt.figure(figsize=(7, 6))

plt.scatter(
    y_test,
    best_predictions,
    alpha=0.6
)

minimum = min(y_test.min(), best_predictions.min())
maximum = max(y_test.max(), best_predictions.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.xlabel("Actual Compressive Strength (MPa)")
plt.ylabel("Predicted Compressive Strength (MPa)")
plt.title(f"Actual vs Predicted: {best_model_name}")
plt.tight_layout()
plt.show()

# 12. Residual analysis

# Residual = Actual value - Predicted value
residuals = y_test.to_numpy() - best_predictions

print("\nRESIDUAL ANALYSIS")
print("Mean residual:", round(np.mean(residuals), 4))
print("Residual standard deviation:", round(np.std(residuals), 4))
print("Residual MAE:", round(np.mean(np.abs(residuals)), 4))

# Residual distribution
plt.figure(figsize=(8, 5))
plt.hist(residuals, bins=25, edgecolor="black")
plt.xlabel("Residual (MPa)")
plt.ylabel("Frequency")
plt.title("Distribution of Prediction Residuals")
plt.tight_layout()
plt.show()

# Residuals against predicted values
plt.figure(figsize=(8, 5))
plt.scatter(best_predictions, residuals, alpha=0.6)
plt.axhline(0, linestyle="--")
plt.xlabel("Predicted Compressive Strength (MPa)")
plt.ylabel("Residual (MPa)")
plt.title("Residuals vs Predicted Strength")
plt.tight_layout()
plt.show()


# Examine whether prediction errors differ across curing-age groups

age_column = next(
    (col for col in X_test.columns if col.lower() == "age"),
    None
)

if age_column is not None:

    age_analysis = pd.DataFrame({
        "Age": X_test[age_column].to_numpy(),
        "Actual": y_test.to_numpy(),
        "Predicted": best_predictions,
        "Residual": residuals
    })

    age_analysis["Age Group"] = pd.cut(
        age_analysis["Age"],
        bins=[0, 7, 28, 90, 365],
        labels=[
            "1-7 days",
            "8-28 days",
            "29-90 days",
            "91-365 days"
        ],
        include_lowest=True
    )

    age_results = age_analysis.groupby(
        "Age Group",
        observed=False
    ).agg(
        Records=("Residual", "count"),
        MAE=("Residual", lambda x: np.mean(np.abs(x))),
        Mean_Residual=("Residual", "mean")
    ).reset_index()

    print("\nERROR ANALYSIS BY AGE")
    display(age_results.round(4))

# 14. Error analysis by actual strength

strength_analysis = pd.DataFrame({
    "Actual": y_test.to_numpy(),
    "Predicted": best_predictions,
    "Residual": residuals
})

strength_analysis["Strength Group"] = pd.qcut(
    strength_analysis["Actual"],
    q=4,
    duplicates="drop"
)

strength_results = strength_analysis.groupby(
    "Strength Group",
    observed=False
).agg(
    Records=("Residual", "count"),
    MAE=("Residual", lambda x: np.mean(np.abs(x))),
    Mean_Residual=("Residual", "mean")
).reset_index()

print("\nERROR ANALYSIS BY ACTUAL STRENGTH")
display(strength_results.round(4))

# 15. CEMENT AND COMPRESSIVE STRENGTH ASSOCIATION
# ------------------------------------------------

if cement_column is not None:

    pearson_corr = df[cement_column].corr(
        df[TARGET],
        method="pearson"
    )

    spearman_corr = df[cement_column].corr(
        df[TARGET],
        method="spearman"
    )

    print("\nCEMENT AND STRENGTH ASSOCIATION")
    print("Pearson correlation:", round(pearson_corr, 4))
    print("Spearman correlation:", round(spearman_corr, 4))

# 16. exporting results

OUTPUT_FOLDER = r"D:\Sheeba\Coursework"

# Export model comparison
results_df.to_csv(
    os.path.join(OUTPUT_FOLDER, "Concrete_Model_Comparison.csv"),
    index=False
)

# Export actual and predicted results
prediction_results = pd.DataFrame({
    "Actual_Strength_MPa": y_test.to_numpy(),
    "Predicted_Strength_MPa": best_predictions,
    "Residual_MPa": residuals
})

prediction_results.to_csv(
    os.path.join(OUTPUT_FOLDER, "Concrete_Predictions.csv"),
    index=False
)

# Export age analysis if available
if age_column is not None:
    age_results.to_csv(
        os.path.join(OUTPUT_FOLDER, "Concrete_Age_Error_Analysis.csv"),
        index=False
    )

# Export strength-group analysis
strength_results.to_csv(
    os.path.join(OUTPUT_FOLDER, "Concrete_Strength_Error_Analysis.csv"),
    index=False
)

# 17. project 2 summary

print("\n" + "=" * 60)
print("CONCRETE COMPRESSIVE STRENGTH PROJECT SUMMARY")
print("=" * 60)

print("Original dataset records: 1030")
print("Records after duplicate removal:", len(df))
print("Number of input features:", X.shape[1])
print("Target variable:", TARGET)
print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print("\nBest model:", best_model_name)

best_mae = mean_absolute_error(y_test, best_predictions)
best_rmse = np.sqrt(
    mean_squared_error(y_test, best_predictions)
)
best_r2 = r2_score(y_test, best_predictions)

print("MAE:", round(best_mae, 4), "MPa")
print("RMSE:", round(best_rmse, 4), "MPa")
print("R2:", round(best_r2, 4))

print("\nRESULT FILES SAVED IN:")
print(OUTPUT_FOLDER)

print("\nPROJECT COMPLETED SUCCESSFULLY!")
```
